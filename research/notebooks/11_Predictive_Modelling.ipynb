# Predictive Modelling
Build a churn prediction model that **maximizes identification of churn-risk customers**, supports **risk segmentation**, and enables **targeted retention actions**, rather than blindly maximizing accuracy.

### Churn Definition

* **Churn = customer leaves the company**

* Business cost of:

  * False Negative (FN) → Customer will actually churn but model predict no churn (HIGH COST)

  * False Positive (FP) → Customer will not churn but model predict churn - To keep those customers company offers unnecessary discount (LOWER COST)


**Therefore:**
Recall for churners matters more than accuracy.

## Find out the best threshold

A typical workflow in production is:

1. Train the model.
2. Predict probabilities (predict_proba).
3. Evaluate the default threshold (0.5).
4. Plot the Precision-Recall curve and ROC curve.
5. Test multiple thresholds.
6. Select the threshold based on the business objective (e.g., maximize recall, maximize F1, or minimize expected business cost).
7. Retrain only if needed; otherwise, deploy the model with the chosen threshold.

````python
## Prediction the probabilites with the best model
y_prob = best_clf.predict_proba(X_test)[:, 1]
y_prob
````

````python
## Draw the Precision-Recall Curve
from sklearn.metrics import precision_recall_curve
import matplotlib.pyplot as plt

precision, recall, thresholds = precision_recall_curve(
    y_test,
    y_prob
)

plt.figure(figsize=(8,6))

plt.plot(recall, precision, linewidth=2)

plt.xlabel("Recall")
plt.ylabel("Precision")
plt.title("Precision-Recall Curve")

plt.grid(True)

plt.show()
````

````python
## Compute the PR-AUC
# The area under the Precision-Recall Curve summarizes performance in a single value.
from sklearn.metrics import average_precision_score

pr_auc = average_precision_score(y_test, y_prob)
print("Average Precision (PR-AUC):", pr_auc)
````

````python
## Draw the ROC Curve
from sklearn.metrics import roc_curve
from sklearn.metrics import roc_auc_score

fpr, tpr, thresholds = roc_curve(
    y_test,
    y_prob
)

roc_auc = roc_auc_score(
    y_test,
    y_prob
)

plt.figure(figsize=(8,6))

plt.plot(
    fpr,
    tpr,
    label=f"AUC = {roc_auc:.3f}",
    linewidth=2
)

plt.plot(
    [0,1],
    [0,1],
    linestyle="--",
    label="Random Classifier"
)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")

plt.title("ROC Curve")

plt.legend()

plt.grid(True)

plt.show()
````

````python
print("ROC-AUC:", roc_auc)
````

````python
## Determine the Best Threshold Using Youden's J Statistic
# A common way to select a threshold from the ROC curve is to maximize TPR − FPR.

J = tpr - fpr
ix = np.argmax(J)

best_threshold = thresholds[ix]

print("Best Threshold:", best_threshold)
print("Sensitivity (TPR):", tpr[ix])
print("False Positive Rate:", fpr[ix])
````

````python
## Final evaluation with thresold 0.41
threshold = 0.41
y_pred = (y_prob>=threshold).astype(int)

print(classification_report(y_true=y_test, y_pred=y_pred))
````

## Interpretation

### 1. Default Threshold (0.50)

Using the default threshold of **0.50**, the model achieved:

* **77% overall accuracy**
* **56% precision** for churn prediction
* **74% recall** for churn prediction
* **F1-score of 0.64**

This means that:

* The model correctly identified **74% of customers who actually churned**.
* About **26% of churning customers were missed**.
* Among all customers predicted as churners, **56% actually churned**.

This provides a balanced baseline model with reasonable precision and recall.

<br/>

### 2. Optimized Threshold (0.41)

After reducing the classification threshold to **0.41**, the model became more sensitive to identifying churning customers.

The updated performance was:

* **75% overall accuracy**
* **51% precision**
* **84% recall**
* **F1-score remained 0.64**

The lower threshold resulted in:

* An increase in churn recall from **74% to 84%**.
* A decrease in churn precision from **56% to 51%**.
* A slight decrease in overall accuracy from **77% to 75%**.

This indicates that the model predicts more customers as potential churners, thereby identifying more actual churners while also increasing the number of false positives.

<br/>

## Business Interpretation

The objective of customer churn prediction is to identify customers who are likely to leave so that the business can take preventive actions.

Comparing the two thresholds:

### Threshold = 0.50

* Detects **74%** of churning customers.
* Misses **26%** of actual churners.
* Produces fewer false alarms.

### Threshold = 0.41

* Detects **84%** of churning customers.
* Misses only **16%** of actual churners.
* Produces more false alarms because precision decreases.

For customer retention campaigns, missing a customer who is about to churn is often more costly than contacting a customer who would have stayed. Therefore, improving recall is usually more valuable than maximizing precision.

<br/>

# Why Accuracy Decreased

The slight reduction in accuracy (77% → 75%) is expected after lowering the threshold.

As the threshold decreases:

* More customers are classified as churners.
* True Positives increase.
* False Positives also increase.

Since churn datasets are imbalanced, **accuracy alone is not an appropriate metric** for model selection. Recall, Precision, F1-score, ROC-AUC, and PR-AUC provide a more meaningful assessment.

<br/>

### Final Conclusion

The threshold optimization successfully shifted the model toward identifying more customers at risk of churning.

Although lowering the threshold from **0.50 to 0.41** reduced overall accuracy and precision slightly, it **improved churn recall from 74% to 84% while maintaining the same F1-score (0.64)**. This means the model is able to identify significantly more potential churners without reducing the overall balance between precision and recall.

From a business perspective, the **0.41 threshold is the preferred operating point** because customer churn prediction prioritizes identifying as many at-risk customers as possible. The increase in false positives is generally acceptable since the cost of offering retention incentives to some customers who would not have churned is often much lower than the cost of losing actual customers.

Therefore, **the model with a threshold of 0.41 is selected as the final deployment model** because it better aligns with the business objective of maximizing customer retention.


## Export the model

````python
import joblib

class ChurnPredictor:

    def __init__(self, model, threshold):
        self.model = model
        self.threshold = threshold

    def predict(self, X):

        prob = self.model.predict_proba(X)[:,1]

        return (prob >= self.threshold).astype(int)

    def predict_proba(self, X):

        return self.model.predict_proba(X)
    

## Save the model
predictor = ChurnPredictor(best_clf, 0.41)

joblib.dump(predictor, "../Artifacts/Model/churn_predictor.pkl")
````


### What Happens in Industry?

In production systems, it's common to keep the threshold in a configuration file or model metadata rather than hard-coding it.

For example:

```yaml
model:
    name: churn_model.pkl
    threshold: 0.41
    version: 1.0.0
```

or

```json
{
    "model": "churn_model.pkl",
    "threshold": 0.41,
    "roc_auc": 0.84,
    "created_at": "2026-07-22"
}
```

The inference service reads the model and the threshold together, ensuring predictions use the same operating point that was selected during evaluation.
